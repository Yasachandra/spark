```python
# Check aggregations on the dataframe
yellowTaxiAnalyzedDF = yellowTaxisDF.describe("passenger_count", "trip_distance")
yellowTaxiAnalyzedDF.show()
```

```python
print("Before operation = " + str(yellowTaxisDF.count()))
# Clean up the data frame by removing records where passenger_count < 1 and and trip_distance < 0
yellowTaxisDF = yellowTaxisDF.where("passenger_count > 0").filter(col("trip_distance") > 0.0)
print("After operation = " + str(yellowTaxisDF.count()))
```

```python
# Remove null values from the data frame
print("Before operation = " + str(yellowTaxisDF.count()))
yellowTaxisDF = yellowTaxisDF.na.drop('all')
print("After operation = " + str(yellowTaxisDF.count()))
```

```python
# Replace null values with default values
defaultValueMap = {'payment_type': 5, 'RateCodeID': 1}
yellowTaxisDF = yellowTaxisDF.na.fill(defaultValueMap)
```

```python
# Remove duplicate records on some specific or all columns
print("Before operation = " + str(yellowTaxisDF.count()))
yellowTaxisDF = yellowTaxisDF.dropDuplicates()
print("After operation = " + str(yellowTaxisDF.count()))
```

```python
# Remove records other than October 2022
print("Before operation = " + str(yellowTaxisDF.count()))
yellowTaxisDF = yellowTaxisDF.where("tpep_pickup_datetime >= '2022-10-01' AND tpep_dropoff_datetime <= '2022-11-01'")
print("After operation = " + str(yellowTaxisDF.count()))
```

```python
yellowTaxisDF.printSchema()
```

```python
yellowTaxisDF = yellowTaxisDF.select("VendorID",
                                     col("passenger_count").cast(IntegerType()),
                                     column("trip_distance").alias("TripDistance"),
                                     yellowTaxisDF.tpep_pickup_datetime,
                                     "tpep_dropoff_datetime",
                                     "PULocationID",
                                     "DOLocationID",
                                     "RatecodeID",
                                     "total_amount",
                                     "payment_type",
)
yellowTaxisDF.printSchema()
```

```python
yellowTaxisDF = yellowTaxisDF.withColumnRenamed("passenger_count", "PassengerCount").withColumnRenamed("tpep_pickup_datetime", "PickupTime").withColumnRenamed("tpep_dropoff_datetime", "DropTime").withColumnRenamed("PULocationID", "PickupLocationID").withColumnRenamed("DOLocationID", "DropLocationID").withColumnRenamed("total_amount", "TotalAmount").withColumnRenamed("payment_type", "PaymentType")
yellowTaxisDF.printSchema()
```

```python
# Create derived columns for year, month and day
yellowTaxisDF = yellowTaxisDF.withColumn("TripYear", year(col("PickupTime"))).select("*", expr("month(PickupTime) as TripMonth"), dayofmonth(col("PickupTime")).alias("TripDay"))
yellowTaxisDF.printSchema()
```

```python
yellowTaxisDF = yellowTaxisDF.withColumn("TripTimeInMinutes", round((unix_timestamp(col("DropTime")) - unix_timestamp(col("PickupTime"))) / 60))
yellowTaxisDF.printSchema()
```

```python
tripTimeInSecondsExpr = unix_timestamp(col("DropTime")) - unix_timestamp(col("PickupTime"))
tripTimeInMinutesExpr = round(tripTimeInSecondsExpr / 60)
yellowTaxisDF = yellowTaxisDF.withColumn("TripTimeInMinutes", tripTimeInMinutesExpr)
yellowTaxisDF.printSchema()
```

```python
tripTypeColumn = when(col("RatecodeID") == 6, "SharedTrip").otherwise("SoloTrip")
yellowTaxisDF = yellowTaxisDF.withColumn("TripType", tripTypeColumn).drop("RatecodeID")
yellowTaxisDF.printSchema()
```

```python
yellowTaxisDFReport = yellowTaxisDF.groupby("PickupLocationID", "DropLocationID").agg(
    sum("TotalAmount").alias("SumAmount"),
    avg("TripTimeInMinutes").alias("AvgTripTime"),
).orderBy(col("PickupLocationId").desc())
yellowTaxisDFReport.show(truncate=False)
```
