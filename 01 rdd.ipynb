```python
from pyspark.sql import SparkSession
from pyspark.sql.types import *
spark = (
    SparkSession
        .builder
        .appName("RDDApp")
        .master("local[4]")
        .getOrCreate()
)
```

```python
sc = spark.sparkContext
```

```python
numbersRdd = sc.parallelize([1, 2, 3, 4, 5])
```

```python
numbersRdd.getNumPartitions() 
```

```python
# Get result from RDD
output = numbersRdd.collect()
print("Output: ", output)
```

```python
# Get 2 records from RDD
numbersRdd.take(2)
```

```python
# Get first record from RDD
firstRecord = numbersRdd.first()
print("First Record: ", firstRecord)
```

```python
evenNumsRdd = numbersRdd.filter(lambda x: x % 2 == 0)
print("Even Numbers: ", evenNumsRdd.collect())
```

```python
# Create RDD with complex types

employeesRdd = sc.parallelize([
    (1, "John", 1000),
    (2, "Jane", 2000),
    (3, "Doe", 3000)
])
```

```python
# Get first record from RDD
firstRecord = employeesRdd.first()
print("First Record: ", firstRecord)
```

```python
# Read TaxiZones.csv file and create RDD
taxiZonesRdd = sc.textFile("./Files/TaxiZones.csv")
```

```python
# Get 10 records from RDD
taxiZonesRdd.take(10)
```

```python
taxiZonesRdd.getNumPartitions()
```

```python
# Read TaxiZones.csv file and create RDD with 4 partitions
taxiZonesRddWith4Partitions = sc.textFile("./Files/TaxiZones.csv", 4)
```

```python
# Apply map operation to split each record by comma, and get array of items
taxiZonesWithColsRdd = (taxiZonesRdd.map (lambda zone: zone.split(",")))
taxiZonesWithColsRdd.take(10)
```

```python
taxiZonesRddWith4Partitions.getNumPartitions()
```

```python
# Apply map operation to split each record by comma, and get array of items
taxiZonesRddWith4Partitions = taxiZonesRddWith4Partitions.map(lambda x: x.split(","))
# Get 10 records from RDD
taxiZonesRddWith4Partitions.take(10)
```

```python
import math

numbersRDD2 = sc.parallelize([2, 3, 4, 5, 6])

# Create pair RDD
numsWithSquareRootRDD = numbersRDD2.map(lambda num: (num, math.sqrt(num)))
numsWithSquareRootRDD.collect()
```

```python
taxiZoneDetailsRDD = taxiZonesWithColsRdd.map(lambda row: (row[0], row[3]))
taxiZoneDetailsRDD.collect()
```

```python
# Create pair RDD (locationId as Key, Value as 1)
taxiZonesPairRDD = (taxiZonesWithColsRdd.map(lambda row: (row[1], 1)))
taxiZonesPairRDD.collect()
```

```python
# Calculate count of records for each locationId
locationCountRDD = (taxiZonesPairRDD.reduceByKey(lambda x, y: x + y))
locationCountRDD.collect()
```

```python
# Sort the data by key
locationCountRDD.sortByKey().collect()
```

```python
# Only return the keys
taxiZonesPairRDD.keys().collect()
```

```python
# Only return the values
taxiZonesPairRDD.values().collect()
```

```python
# Return distinct pairs
taxiZonesPairRDD.distinct().collect()
```
